```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("train.csv")


df.info() 
df.describe() 
plt.xlabel('df.Survived')
plt.ylabel('df.Sex')
plt.plot(df.Survived,df.Sex)
```

```python
df.groupby("Sex")["Survived"].mean()
sns.barplot(
    data = df,
    x = "Sex",
    y = "Survived"
)
plt.xlabel("Gender")
plt.ylabel("Survived") 
plt.show() 
```
